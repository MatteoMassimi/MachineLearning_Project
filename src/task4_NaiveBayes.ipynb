## 0. Setup 

In questo primo step , vengono importate le librerie e il dataset `training.csv`.


```python
import pandas as pd
import numpy as np
```

```python
training = pd.read_csv("../data/processed/training.csv", sep=";")
dataFrame_NB = training.copy() 
```

## 1. Richiamo di variabili e metodi utilizzati in Naive Bayes
In questa sezione riprendiamo variabili e metodi necessati per testare Naive Bayes su `training.csv`.
In particolare andremo a riprendere **predici_naive_bayes** ,che implementa il modello vero e proprio, e **probabilita**. 



```python
#Tutte le probabilità condizionate calcolate con lo stimatore di Laplace.

probabilita = {
    "marital": {
        0: {"divorced": 1/9, "married": 7/9, "single": 1/9},
        1: {"divorced": 1/6, "married": 2/6, "single": 3/6},
    },
    "housing": {
        0: {"no": 2/6, "yes": 4/6},
        1: {"no": 2/6, "yes": 4/6},
    },
    "loan": {
        0: {"no": 5/6, "yes": 1/6},
        1: {"no": 7/8, "yes": 1/8},
    }
}
```

```python
def predici_naive_bayes(row):
    # Probabilità a priori calcolate nel task 2
    score_0 = 0.5
    score_1 = 0.5

    # marital
    score_0 *= probabilita["marital"][0][row["marital"]] 
    score_1 *= probabilita["marital"][1][row["marital"]]

    # housing
    score_0 *= probabilita["housing"][0][row["housing"]]
    score_1 *= probabilita["housing"][1][row["housing"]]

    # loan
    score_0 *= probabilita["loan"][0][row["loan"]]
    score_1 *= probabilita["loan"][1][row["loan"]]



    # Predizione finale 
    if (score_0 > score_1):
        prediction = 0
    else:
        prediction = 1
    return prediction, score_0, score_1
```

## 2. Naïve Bayes su `training.csv`

Una volta ripreso cio che ci serve , possiamo testare il modello su `training.csv`.

```python
dataFrame_NB["Predicted"] = dataFrame_NB.apply(lambda row: predici_naive_bayes(row)[0],axis=1)
dataFrame_NB[["y", "Predicted"]].head(10)
```

## 3. Valutazione delle prestazioni 

Una volta applicato il modello sul dataset `training.csv`, possiamo valutare le prestazioni basandoci sui seguenti criteri :

- Confusion Matrix
- Accuracy
- Precision
- Recall
- F1

I parametri utilizzati per calcolarci tali metriche di valutazione sono :

- TP = True Positive
- TN = True Negative
- FP = False Positive
- FN = False Negative



## 3.1 Confusion Matrix

```python

tp = len(dataFrame_NB[(dataFrame_NB["y"] == 1) & (dataFrame_NB["Predicted"] == 1)])
tn = len(dataFrame_NB[(dataFrame_NB["y"] == 0) & (dataFrame_NB["Predicted"] == 0)])
fp = len(dataFrame_NB[(dataFrame_NB["y"] == 0) & (dataFrame_NB["Predicted"] == 1)])
fn = len(dataFrame_NB[(dataFrame_NB["y"] == 1) & (dataFrame_NB["Predicted"] == 0)])


confusion_matrix_df = pd.DataFrame(
    [[tn, fp],
     [fn, tp]],
    columns=["Predetto 0", "Predetto 1"],
    index=["Reale 0", "Reale 1"]
)
print("TP = ", tp)
print("TN = ", tn)
print("FP = ", fp)
print("FN = ", fn)
print("\nMatrice di Confusione:")
confusion_matrix_df

```

## 3.2 Accuracy

L'**accuracy** indica quanto e stato bravo il modello.

```python
accuracy = (dataFrame_NB["y"] == dataFrame_NB["Predicted"]).mean()
print(f"Accuracy: {accuracy:.4f}")
```

## 3.3 Precision

La **precision** indica in percentuale quanti tra tutti i positivi individuati sono realmente positivi.

$$ \text{Precision} = \frac{TP}{TP + FP} $$

```python
precision = tp / (tp + fp)
print(f"Precision: {precision:.4f}")
```

## 3.4 Recall

La **recall** indica in percentuale quanti positivi sono stati individuati

$$ \text{Recall} = \frac{TP}{TP + FN} $$

```python
recall = tp / (tp + fn)
print(f"Recall: {recall:.4f}")
```

## 3.5 F1

L' **F1** rappresenta la media ponderata tra precision e recall.

$$ F1 = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}} $$

```python
f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0.0
print(f"F1 Score: {f1:.4f}")
```

L'accuracy risulta moderata, ma non rappresenta da sola una misura affidabile delle prestazioni del classificatore, poiché il dataset presenta un forte sbilanciamento tra le classi.

La recall risulta inferiore al 50% (45.18%), cio mosta che il modello individua discretamente i reali positivi.

### 5. Ottimizzazione

Avendo scelto di valutare il modello senza riaddestrarlo , il margine di
ottimizzazione disponibile in questo task è **limitato**: 

Qui abbiamo solo cambiato le probabilita a priori basandoci sui risultati ottenuti nel task 1.

Le leve di ottimizzazione vere e proprie come ad esempio  **riaddestrare** i modelli su `training.csv`,
**aggiungere attributi** più informativi, **discretizzare** le variabili numeriche, e soprattutto
**gestire lo sbilanciamento** delle classi (es. con pesi di classe o ricampionamento), 
 vengono trattate nel**Task 5**.


```python
def predici_naive_bayes_ottimizzato(row):
    # Probabilità a priori cambiate in base alla distribuzione reale del target nel training set
    score_0 = 0.887
    score_1 = 0.113

    # marital
    score_0 *= probabilita["marital"][0][row["marital"]] 
    score_1 *= probabilita["marital"][1][row["marital"]]

    # housing
    score_0 *= probabilita["housing"][0][row["housing"]]
    score_1 *= probabilita["housing"][1][row["housing"]]

    # loan
    score_0 *= probabilita["loan"][0][row["loan"]]
    score_1 *= probabilita["loan"][1][row["loan"]]

    # Predizione finale 
    if (score_0 > score_1):
        prediction = 0
    else:
        prediction = 1
    return prediction, score_0, score_1
```

```python
dataFrame_NB["Predicted"] = dataFrame_NB.apply(lambda row: predici_naive_bayes_ottimizzato(row)[0],axis=1)
dataFrame_NB[["y", "Predicted"]].head(10)
```

## 6. Valutazione prestazioni modello ottimizzato

Di seguito verranno ricalcolate tutte le metriche basandosi sulle nuove probabilita a priori

```python
tp_opt = len(dataFrame_NB[(dataFrame_NB["y"] == 1) & (dataFrame_NB["Predicted"] == 1)])
tn_opt = len(dataFrame_NB[(dataFrame_NB["y"] == 0) & (dataFrame_NB["Predicted"] == 0)])
fp_opt = len(dataFrame_NB[(dataFrame_NB["y"] == 0) & (dataFrame_NB["Predicted"] == 1)])
fn_opt = len(dataFrame_NB[(dataFrame_NB["y"] == 1) & (dataFrame_NB["Predicted"] == 0)])


confusion_matrix_df = pd.DataFrame(
    [[tn_opt, fp_opt],
     [fn_opt, tp_opt]],
    columns=["Predetto 0", "Predetto 1"],
    index=["Reale 0", "Reale 1"]
)
accuracy = (dataFrame_NB["y"] == dataFrame_NB["Predicted"]).mean()
precision = tp_opt / (tp_opt + fp_opt) if( tp_opt + fp_opt ) > 0 else 0.0
recall = tp_opt / (tp_opt + fn_opt) if (tp_opt + fn_opt) > 0 else 0.0
f1_score = 2 * (precision * recall) / (precision + recall) if  (precision + recall) > 0 else 0.0

print("TP = ", tp_opt)
print("TN = ", tn_opt)
print("FP = ", fp_opt)
print("FN = ", fn_opt)
print("Accuracy = ", accuracy)
print("Precision = ", precision)
print("Recall = ", recall)
print("F1 Score = ", f1_score)
print("\nMatrice di Confusione:")
confusion_matrix_df
```

## 7. Confronto tra modello originale e modello ottimizzato

Dopo aver applicato il classificatore Naive Bayes manuale al dataset training.csv, sono state calcolate le metriche di performance sia per il modello originale (basato sulle probabilità stimate da manuale.csv) sia per il modello ottimizzato (con probabilità a priori ricalcolate sull intero set).

| Metrica | Modello Originale | Modello Ottimizzato |
|---|---|---|
| Accuracy  | 59.63% | 88.73%|
| Precision | 12.96% | 0%|
| Recall    | 45.18% | 0%|
| F1-Score  | 20.14% | 0%|


#### **Considerazioni**

A un primo sguardo l'ottimizzazione sembra un grande successo: l'accuracy passa dal **59.63%** all'**88.73%**, quasi 30 punti in più. In realtà si tratta di un **miglioramento illusorio**, e l'analisi delle altre metriche lo dimostra chiaramente.

**1. L'accuracy alta è un artefatto dello sbilanciamento.** Sostituendo le probabilità a priori bilanciate (0.5 / 0.5) con quelle reali del dataset (≈ 0.887 / 0.113, stimate nel Task 1), il termine a priori della classe `0` diventa così dominante che le tre probabilità condizionate di `marital`, `housing` e `loan` non riescono mai a ribaltarlo: lo `score_1` non supera **mai** lo `score_0`. Il risultato è un modello **degenere** che predice `0` per *tutte* le 41.176 istanze (FP = 0 e **TP = 0**). L'88.73% di accuracy non misura quindi una reale capacità predittiva: coincide esattamente con la **proporzione della classe maggioranza** nel dataset, cioè con ciò che otterrebbe un banale classificatore *Zero-R* che risponde sempre "no". È il classico **paradosso dell'accuracy** su dati sbilanciati.

**2. Precision, Recall e F1 a zero rivelano l'inutilità del modello ottimizzato.** Non predicendo mai la classe positiva, il modello non individua **nessuno** dei 4.639 clienti che hanno effettivamente sottoscritto (recall = 0). Proprio per l'obiettivo di business — riconoscere i potenziali sottoscrittori, che sono l'evento *raro* di interesse — il modello "ottimizzato" è completamente **inservibile**, nonostante l'accuracy più alta. Questo conferma quanto osservato nel Task 2 e nel Task 3: su un problema fortemente sbilanciato la sola accuracy è **fuorviante**, e vanno guardate precision, recall e F1 sulla classe positiva.

**3. Il vero trade-off tra i due modelli.** Il modello originale, pur con un'accuracy mediocre, ha almeno il pregio di *provare* a individuare i positivi: con priori bilanciate recupera il **45%** dei sottoscrittori (recall 45.18%), al prezzo di molti falsi positivi (precision 12.96%) e quindi di un'accuracy bassa. Il modello ottimizzato fa l'opposto estremo: massimizza l'accuracy azzerando però del tutto la copertura dei positivi. Nessuno dei due è soddisfacente, ma il **secondo è chiaramente peggiore** rispetto allo scopo dell'analisi, perché rinuncia completamente alla classe che vogliamo prevedere.

**4. Perché il margine di ottimizzazione qui è strutturalmente limitato.** Come anticipato nella Sezione 5, in questo task il modello **non viene riaddestrato**: l'unica leva disponibile è la modifica delle probabilità a priori, mentre le probabilità condizionate restano quelle stimate sulle sole 12 istanze di `manuale.csv`. Inoltre il classificatore poggia su appena **tre feature** (`marital`, `housing`, `loan`), due delle quali — `housing` e `loan` — si erano già rivelate **poco discriminanti** nel Task 2 (distribuzioni quasi identiche nelle due classi). Con predittori così deboli, agire solo sui priori non può che produrre i due estremi visti sopra: o troppi falsi positivi, o nessun positivo.

**Conclusione.** Il confronto mostra che intervenire solo sulle probabilità a priori **non è una vera ottimizzazione**: sposta semplicemente il modello lungo il trade-off precision/recall senza migliorarne il potere discriminante. Le leve realmente efficaci — **riaddestrare** i modelli direttamente su `training.csv`, introdurre **attributi più informativi** (es. `poutcome`, `contact`, `month`, emersi come rilevanti nel Task 3), **discretizzare** le variabili numeriche ed escludere quelle problematiche (`duration`, `pdays`), e soprattutto **gestire esplicitamente lo sbilanciamento** delle classi (pesi di classe o ricampionamento) — esulano dalla logica "a mano" di questo task e vengono affrontate nel **Task 5**, dove la differenza tra i due classificatori potrà essere valutata in modo statisticamente fondato.

---

> **Nota — perché Naive Bayes e 1R danno gli stessi risultati.** Confrontando questo notebook con `task4_1R.ipynb` si osserva che i due classificatori producono **esattamente le stesse metriche** (sia il modello originale sia quello ottimizzato). **Non è una coincidenza**, ma una conseguenza diretta delle feature scelte nel Task 2. La decisione di Naive Bayes è $\arg\max_c\ P(c)\cdot P(\text{marital}\mid c)\cdot P(\text{housing}\mid c)\cdot P(\text{loan}\mid c)$, e in questo confronto:
>
> - i **priori** sono uguali (0.5 / 0.5) e quindi **si elidono**;
> - **`housing`** ha distribuzioni **identiche** nelle due classi ($P(\cdot\mid 0)=P(\cdot\mid 1)$): il suo fattore è lo stesso per `score_0` e `score_1` e **si semplifica**, senza influenzare la decisione;
> - **`loan`** è quasi simmetrico e comunque **troppo debole**: il suo contributo non basta mai a ribaltare la classe suggerita da `marital`.
>
> Di fatto la predizione è guidata dal **solo `marital`**, esattamente l'attributo su cui 1R costruisce la propria regola: ecco perché i due modelli prendono le **stesse** decisioni e ottengono le **stesse** metriche. È un risultato **atteso e spiegabile**, non un errore: mostra come un Naive Bayes costruito su poche feature poco discriminanti **degeneri** in un classificatore a regola singola, indistinguibile da 1R. La differenza tra i due approcci emergerà nel **Task 5**, dove Naive Bayes userà *tutte* le feature, verrà riaddestrato e avrà un preprocessing completo.
